###Q3 Write a Python program to add, subtract, multiple and divide two Pandas Series.

```python
ar1=[2, 4, 6, 8, 10]
ar2=[1, 3, 5, 7, 9]
d1=pd.Series(ar1)
d2=pd.Series(ar2)
print("Add")
print(d1+d2)
print("Subtract")
print(d1-d2)
print("Multiply")
print(d1*d2)
print("Divide")
print(d1/d2)
```
