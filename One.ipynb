###Q12 Write a Python program to find common values between two arrays.

```python
arr=[]
ar1 = np.array([0 ,10 ,20 ,40, 60])
ar2 = np.array([10, 30, 40])
for i in range (0,ar1.size):
    if ar1[i] in ar2:
        arr.append(ar1[i])
arr=np.unique(arr)
arr
```

###Q13 Write a Python program to find the set difference of two arrays. The set difference
will return the sorted, unique values in array1 that are not in array2.

```python
arr=[]
ar1=np.array( [ 0 ,10 ,20 ,40 ,60 ,80])
ar2=np.array([10, 30, 40, 50, 70, 90])
for i in range (0,ar1.size):
    if ar1[i] not in ar2:
        arr.append(ar1[i])
arr=np.sort(arr)
arr=np.unique(arr)
arr
```

###Q14 Write a Python program to find the set exclusive-or of two arrays. Set exclusive-or
