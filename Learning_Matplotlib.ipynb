#Learning Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
```

```python
x=[0,1,2,3,4]
y=[0,2,4,6,8]
plt.figure(figsize=(5,3),dpi=1000)

#plt.plot(x,y,label='2x',color='red',linewidth=2,linestyle='--',marker='.',markersize=10,markeredgecolor='blue')
plt.plot(x,y,'b^--',label='2x')


x2=np.arange(0,4.5,0.5)
plt.plot(x2[:5],x2[:5]**2,'r',label='x^2')
plt.plot(x2[4:],x2[4:]**2,'r--',label='x^2')



plt.title("Our Graph",fontdict={'fontname':'Comic Sans MS'})
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.xticks([0,1,2,3])
plt.yticks([0,2,4,6,8,10])

plt.legend()
plt.savefig('mygraph.png')

plt.show()
```

###Bar Chart

```python
labels=['A','B','C']
values=[1,4,2]
bars=plt.bar(labels,values)

bars[0].set_hatch('/')
bars[1].set_hatch('o')
bars[2].set_hatch('*')

plt.figure(figsize=(6,4))


plt.show()
```

### Fifa Data

```python
fifa=pd.read_csv('fifa_data.csv')
fifa.head(5)
```

###Histogram

```python
bins=[40,50,60,70,80,90,100]

plt.hist(fifa.Overall,bins=bins,color='red')
plt.xticks(bins)

plt.ylabel('Number of Players')
plt.xlabel('Skills')
plt.title('FIFA 2018')


plt.show()
```

###Pie Chart

```python
left=fifa.loc[fifa['Preferred Foot']=='Left'].count()[0]
right=fifa.loc[fifa['Preferred Foot']=='Right'].count()[0]
labels=['Left','Right']
colors=['Red','Blue']

plt.pie([left,right],labels=labels,colors=colors,autopct='%.2f %%')


plt.title('Foot Preference')
plt.show()
```

```python
fifa.Weight=[int(x.strip('lbs')) if type(x)==str else x for x in fifa.Weight]

plt.style.use('ggplot')

light=fifa.loc[fifa.Weight<125].count()[0]
light_medium=fifa.loc[(fifa.Weight>=125) & (fifa.Weight<150) ].count()[0]
medium=fifa.loc[(fifa.Weight>=150) & (fifa.Weight<175) ].count()[0]
medium=fifa.loc[(fifa.Weight>=150) & (fifa.Weight<175) ].count()[0]
medium_heavy=fifa.loc[(fifa.Weight>=175) & (fifa.Weight<200) ].count()[0]
heavy=fifa.loc[fifa.Weight>=200].count()[0]


weights=[light,light_medium,medium,medium_heavy,heavy]
labels=["Under - 125" , '15-150','150-175','175-200','Over-200']
explode=(.4,.1,0,0,.4)

plt.title('Fifa Weight')


plt.pie(weights,labels=labels,autopct='%.2f %%',explode=explode)



plt.show()
```

###Box And Wisker Chart

```python
plt.style.use('default')

plt.figure(figsize=(5,8))

barcelona=fifa.loc[fifa.Club == 'FC Barcelona']['Overall']
madrid=fifa.loc[fifa.Club == 'Real Madrid']['Overall']
Revolution=fifa.loc[fifa.Club == 'New England Revolution']['Overall']

labels=['FC Barcelona','Real Madrid','NE Revolution']
boxes=plt.boxplot([barcelona,madrid,Revolution],labels=labels,patch_artist=True,medianprops={'linewidth':2})
for box in boxes['boxes']:
    box.set(color='#4286f4',linewidth=2)

    box.set(facecolor='#e0e0e0')

plt.title("Team Comparison")
plt.ylabel("FIfa Overall Rating")
plt.show()
```
